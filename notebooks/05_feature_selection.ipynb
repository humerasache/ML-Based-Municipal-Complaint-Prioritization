# Feature Selection

## Import Libraries

```python
import pandas as pd
import numpy as np
```

## Load  Data

```python
df = pd.read_csv("../data/processed/feature_engineered_complaints.csv")
```

## Target Variable

```python
target = "severity"

print(df[target].unique())
```

```python
print("Severity distribution:")
display(df[target].value_counts())
```

```python
print("\nSeverity distribution in percentage:")
display(
    (df[target].value_counts(normalize=True) * 100)
    .round(2)
)
```

## Separate Features and Target

```python
X = df.drop(columns=[target])
Y = df[target]
```

```python
print("Features shape:", X.shape)
print("Target shape:", Y.shape)
```

```python
print("\nTarget classes:")
print(sorted(Y.unique()))
```

## Columns for ML Training

```python
columns_to_exclude = [
    "complaint_id",
    "citizen_id",
    "ward",
    "reported_date",
    "status",
    "complaint_age_days",
    "is_active_complaint",
    "ward_complaint_load",
]
```

```python
columns_to_exclude = [
    col for col in columns_to_exclude
    if col in X.columns
]

X = X.drop(columns=columns_to_exclude)
```

```python
print("Excluded columns:", columns_to_exclude)
print("\nRemaining feature columns:")
print(X.columns)
```

## Categorical Features

```python
categorical_features = X.select_dtypes(
    include=["object", "category", "str"]
).columns

for col in categorical_features:
    print(f"\n{col} by severity:")
    display(pd.crosstab(df[col], Y, normalize="index").round(2))
```

## Save Data

```python
ml_ready_df = X.copy()
ml_ready_df["severity"] = Y
```

```python
ml_ready_df.to_csv("../data/processed/ml_ready_complaints.csv", index=False)
```
